```python
class Employee:

    def __init__(self, name, sal, last):
        self.name = name  # instance variables
        self.__sal = sal   # private variables
        self.last = last 

    def sal_per_annum(self):
        return self.sal * 12
```

```python
obj1 = Employee('raj', 75000, 'sharma') # instance
```

```python
obj1.sal
```

```python
obj1.__sal
```

```python
"""  
create a class called Student
attributes -> name, klass, skills, address(city)

skills -> this will be a list of skills and every student object will initially have no skills
address -> private variable

create init method
create a method which will add the skills in the skills attribute
create a method to see all the info of the student

create 3 object and test the working
""";
```

```python
class Student:

    """ 
    simple student record

    args:
        name: name of the student
        klass: class of the student. eg.. BE, TE
        address: address as city of the student
        
    """

    def __init__ (self, name: str, klass: str, address: str) -> None:
        self.name = name
        self.klass = klass
        self.__address = address
        self.skills = []

    def add_skill(self, s: str) -> list:
        """ 
        args: 
            s: skill to add 

        returns:
            list having the skills appended
        """
        
        return self.skills.append(s)

    def student_info(self) -> str:

        """ 
        returns all the info of the student
        """

        return f"name: {self.name} --- class: {self.klass} --- address: {self.__address} --- skills: {self.skills}"
```

```python
stu1 = Student('raj', 'BE', "Pune")
stu2 = Student('sia', 'BE', "Pune")
stu3 = Student('pravin', "SE", "Nasik")
```

```python
stu1.student_info()
```

```python
stu1.add_skill('java')
```

```python
stu1.student_info()
```

```python
stu1.add_skill('dart')
```

```python
stu1.student_info()
```

```python
"""  
menu driven programm -> Grocery List
add items, view items

create a class for this

so that when u create an object of this class and call a method on it, it should start the app
(starting with showing menu and then u can interact with it)

""";
```

```python
class Menuapp():

    def __init__(self):
        self.itemlist = []

    def add_item(self):
        item = input("add the item ")
        return self.itemlist.append(item)

    def view_item(self):
        print(self.itemlist)

    def run(self):
        while True:
            print("MENU\n1. Add Item\n2. View List\n3. Exit")

            ch = int(input("Enter the choice"))

            if ch == 1:
                self.add_item()
            elif ch == 2:
                self.view_item()
            elif ch == 3:
                break
            else:
                print("Invalid choice")

        
```

```python
menu1 = Menuapp()
```

```python
menu1.run()
```

```python
# getter and setter with encapsulation
```

```python
class Person:

    def __init__(self, name, age) -> None:
        self.__name = name
        self.__age = age

    # getter 
    def get_name(self):
        return self.__name

    # setter
    def set_name(self, name):
        self.__name = name

    def __str__(self):
        return f"name -> {self.__name}___ age -> {self.__age}"
```

```python
o1 = Person('sia', 26)
```

```python
o1.get_name()
```

```python
o1.set_name('siya')
```

```python
o1.get_name()
```

```python
print(o1)  # __str__
```
