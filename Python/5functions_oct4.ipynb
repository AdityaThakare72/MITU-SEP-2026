```python
def sort_list(lis: list[int], order: bool) -> list:
    """  
    takes the list and the boolean value and returns the sorted list

    Args:
        lis: list which u want to sort
        order: boolean value. False -> Ascending, True -> Desc

    Returns:
        the sorted list according the order value
    
    
    """

    l = sorted(lis , reverse= order)
    return l
```

```python
help(sort_list)
```
