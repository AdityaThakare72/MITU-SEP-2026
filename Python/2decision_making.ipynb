```python
# take selling price and purchase price. check if profit or loss and how much

purchase_price = int(input("enter the purchase price"))
selling_price = int(input("enter the selling price"))

if selling_price > purchase_price:
    print("profit of:", selling_price - purchase_price)
else:
    print(f"loss of {purchase_price - selling_price}")
```

```python
year = int(input("enter the year"))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("it is a century leap year")
        else:
            print("century year but not leap")
    else:
        print("not century but leap year")
else:
    print('not leap')
```

```python
""" 
build an ATM machine menu driven program

start with initial balance -> 20000

should be able to perfrom following operations:
check balance
deposit
withdraw

""";
```

```python
initial_balance = 20000
```

```python
print("Welcome to this ATM Machine 💰")

print("what would you like to do:\n1. check balance\n2. deposit\n3. withdraw")

ch = int(input("select from the options provided above"))

if ch == 1:
    print(f"current balance is: {initial_balance}")
elif ch == 2:
    dep = int(input("how much u want to deposite"))
    initial_balance += dep
    print(f"after deposite ur balance is {initial_balance}")
elif ch == 3:
    withdraw = int(input("how much u want to withdraw"))
    if initial_balance >= withdraw:
        initial_balance -= withdraw  # intial_balance = initial_balance - withdraw
        print(f"withdraw successful. ur current balance is: {initial_balance}")
    else:
        print("insufficient balance. please deposite first. UDHAR BAND!!!!")
else:
    print("👓")
```
