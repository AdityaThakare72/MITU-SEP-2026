```python
class Employee:

    def __init__(self, name, sal, last):
        self.name = name  # instance variables
        self.sal = sal
        self.last = last 

    def sal_per_annum(self):
        return self.sal * 12
```

```python
obj1 = Employee('raj', 75000, 'sharma') # instance
```

```python
obj1.name
```

```python
obj1.sal_per_annum()
```

```python
## encapsulation
```

```python
class Employee:

    def __init__(self, name, sal, last):
        self.name = name  # instance variables
        self.__sal = sal   # private variables
        self.last = last 

    def sal_per_annum(self):
        return self.sal * 12
```

```python
obj1 = Employee('raj', 75000, 'sharma') # instance
```

```python
obj1.sal
```

```python
obj1.__sal
```

```python
"""  
create a class called Student
attributes -> name, klass, skills, address(city)

skills -> this will be a list of skills and every student object will initially have no skills
address -> private variable

create init method
create a method which will add the skills in the skills attribute
create a method to see all the info of the student

create 3 object and test the working
""";
```

```python
class Student:

    def __init__ (self, name, klass, address):
        self.name = name
        self.klass = klass
        self.__address = address
        self.skills = []

    def add_skill(self, s):
        return self.skills.append(s)

    def student_info(self):
        print(f"name: {self.name}\nclass: {self.klass}\naddress: {self.__address}\n skills: {self.skills}")
```

```python
stu1 = Student('raj', 'BE', "Pune")
stu2 = Student('sia', 'BE', "Pune")
stu3 = Student('pravin', "SE", "Nasik")
```

```python
stu1.student_info()
```

```python
stu1.add_skill('java')
```

```python
stu1.student_info()
```

```python
stu1.add_skill('dart')
```

```python
stu1.student_info()
```
