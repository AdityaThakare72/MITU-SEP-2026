```python
# printing a name 10 times
count = 1

while count <= 10:
    print("Pravin")
    count = count + 1
```

```python
# print the numbers from 5 to 25 using while loop
```

```python
count = 5

while count <= 25:
    print(count, end= " ")
    count += 1
```

```python
# print odd numbers from 5 to 26

count = 5

while count <= 25:
    print(count, end= " ")
    count += 2
```

```python
# find the addition from 1 to 10 numbers

count = 1

add = 0

while count <= 10:
    add = add + count
    count += 1
    
print(f"the addition of 1 to 10 numbers is {add}")
```
