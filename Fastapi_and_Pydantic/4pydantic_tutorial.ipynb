```python
from pydantic import BaseModel
```

```python
# create a schema (assume we will create a function later on to check name and level)

class Trainers(BaseModel):

    name: str
    level: int
```

```python
# lets create the pydantic object

trainer_info = {'name': 'grey', 'level': '58'}


trainer1 = Trainers(**trainer_info)
```

```python
def trainer_insert(trainer: Trainers):

    print(trainer.name)
    print(trainer.level)
    print('inserted')
```

```python
trainer_insert(trainer1)
```

```python
from typing import List, Dict, Optional
```

```python
class Trainers(BaseModel):

    name: str   # when u add fields like this it will be required
    level: int = 10   # default value
    skills: List[str] 
    region: Optional[str]
```

```python
trainer_info = {'name': 'grey', 'level': '58', 'skills': ['java', 'python'], 'region': "USA"}
```

```python
trainer1 = Trainers(**trainer_info)
```

```python
trainer_insert(trainer1)
```
