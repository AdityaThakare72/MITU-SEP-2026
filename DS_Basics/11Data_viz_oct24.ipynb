```python
def filter_outliers(df):

    for i in df.columns:
        q3 = np.percentile(df[i], 75)
        q1 = np.percentile(df[i], 25)

        iqr = q3 - q1

        ub = q3 + 1.5 * iqr 
        lb = q1 - 1.5 * iqr

        df = df[(df[i] > lb) & (df[i] < ub)]

    return df
    
```
