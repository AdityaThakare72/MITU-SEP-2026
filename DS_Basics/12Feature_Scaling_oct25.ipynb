```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Social_Network_Ads.csv
# url -> https://mitu.co.in/dataset
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
# plot the scatterchart to check the distri of Age and Estimated
# separate Age and estimated salary from the data
```

```python
x = df[['Age', 'EstimatedSalary']]
x.describe()
```

```python
plt.title("age vs exp sal")
plt.xlabel('age')
plt.ylabel('sal')

plt.scatter(x['Age'], x['EstimatedSalary']);
```

```python
# normalization
```

```python
from sklearn.preprocessing import MinMaxScaler   # import the class

# create an object
sca = MinMaxScaler()
```

```python
# get the scaled values

x_scaled = sca.fit_transform(x)
```

```python
x_scaled
```

```python
pd.DataFrame(x_scaled).describe()
```

```python
# plot the scatter chart and check if the distri is same
```

```python
plt.title("age vs exp sal")
plt.xlabel('age')
plt.ylabel('sal')

plt.scatter(x_scaled[:, 0], x_scaled[:, 1]);
```
