```python
gender = ['m', 'm', 'm', 'm', 'f', 'f', 'm', 'f', 'm', 'm']
```

```python
df['cgpa'] = df['marks'] / 10 + 0.5
```

```python
# adding a column if the student is in dist or not based on the marks -> is_dist
# marks > 60 -> Yes, < 60 -> No
```

```python
df['is_dist'] = df['marks'].apply(lambda x: 'yes' if x >= 60 else 'no')
```
